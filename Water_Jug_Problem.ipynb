```python
# prompt: • Solve the Water-Jug Problem using A* Algorithm
# You are given two jugs with capacities of 4 liters and 3 liters. The goal is to measure exactly
# 2 liters using these jugs. Implement the A* algorithm to solve the water-jug problem.
# The possible actions are:
# 1. Fill either jug completely.
# 2. Empty either jug.
# 3. Pour water from one jug into another until one jug is full or the other is empty.
# Use a heuristic that minimizes the absolute difference between the current amount of water
# in either jug and the target.

import heapq

class Node:
    def __init__(self, jug1, jug2, parent=None, action=None):
        self.jug1 = jug1
        self.jug2 = jug2
        self.parent = parent
        self.action = action
        self.cost = 0  # Initialize cost to 0, will be updated during search

    def __lt__(self, other):
        return (abs(self.jug1 - 2) + abs(self.jug2 - 0)) < (abs(other.jug1 - 2) + abs(other.jug2 - 0))

def get_neighbors(node, capacity1, capacity2):
    neighbors = []

    # Fill jug1
    neighbors.append(Node(capacity1, node.jug2, node, "Fill jug1"))

    # Fill jug2
    neighbors.append(Node(node.jug1, capacity2, node, "Fill jug2"))

    # Empty jug1
    neighbors.append(Node(0, node.jug2, node, "Empty jug1"))

    # Empty jug2
    neighbors.append(Node(node.jug1, 0, node, "Empty jug2"))

    # Pour jug1 into jug2
    pour_amount = min(node.jug1, capacity2 - node.jug2)
    neighbors.append(Node(node.jug1 - pour_amount, node.jug2 + pour_amount, node, "Pour jug1 into jug2"))

    # Pour jug2 into jug1
    pour_amount = min(node.jug2, capacity1 - node.jug1)
    neighbors.append(Node(node.jug1 + pour_amount, node.jug2 - pour_amount, node, "Pour jug2 into jug1"))

    return neighbors

def solve_water_jug(capacity1, capacity2, target):
    start_node = Node(0, 0)
    priority_queue = [(start_node.cost + abs(start_node.jug1 - target) + abs(start_node.jug2 - 0), start_node)]
    visited = set()

    while priority_queue:
        _, current_node = heapq.heappop(priority_queue)

        if current_node.jug1 == target and current_node.jug2 == 0:
            path = []
            while current_node:
                path.insert(0, (current_node.jug1, current_node.jug2, current_node.action))
                current_node = current_node.parent
            return path

        if (current_node.jug1, current_node.jug2) in visited:
            continue
        visited.add((current_node.jug1, current_node.jug2))

        for neighbor in get_neighbors(current_node, capacity1, capacity2):
            neighbor.cost = current_node.cost + 1
            heapq.heappush(priority_queue, (neighbor.cost + abs(neighbor.jug1 - target) + abs(neighbor.jug2 - 0), neighbor))

    return None  # No solution found


if __name__ == "__main__":
    capacity1 = 4
    capacity2 = 3
    target = 2

    solution = solve_water_jug(capacity1, capacity2, target)

    if solution:
        print("Solution found:")
        for jug1, jug2, action in solution:
            print(f"Jug1: {jug1}, Jug2: {jug2} ({action})")
    else:
        print("No solution found.")
```
