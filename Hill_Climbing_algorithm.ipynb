Implement Hill-Climbing algorithm to solve the 8-queen problem

```python


import random

def calculate_conflicts(board):
  """Calculates the number of conflicts between queens on the board."""
  conflicts = 0
  for i in range(len(board)):
    for j in range(i + 1, len(board)):
      if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
        conflicts += 1
  return conflicts

def hill_climbing_8queens():
  """Solves the 8-queens problem using the hill-climbing algorithm."""

  board = [random.randint(0, 7) for _ in range(8)]  # Initialize a random board
  current_conflicts = calculate_conflicts(board)

  while current_conflicts > 0:
    best_neighbor = board[:]  # Copy the current board
    best_neighbor_conflicts = current_conflicts

    for i in range(8):
      for j in range(8):
        neighbor = board[:]
        neighbor[i] = j  # Try moving the queen at position i to column j
        neighbor_conflicts = calculate_conflicts(neighbor)

        if neighbor_conflicts < best_neighbor_conflicts:
          best_neighbor = neighbor[:]
          best_neighbor_conflicts = neighbor_conflicts

    if best_neighbor_conflicts < current_conflicts:
      board = best_neighbor
      current_conflicts = best_neighbor_conflicts
    else:
      # Local minimum reached, restart with a new random board
      board = [random.randint(0, 7) for _ in range(8)]
      current_conflicts = calculate_conflicts(board)

  return board

if __name__ == "__main__":
  solution = hill_climbing_8queens()
  print("Solution found:")
  for row in range(8):
    line = ""
    for col in range(8):
      if solution[row] == col:
        line += "Q "
      else:
        line += ". "
    print(line)
```
